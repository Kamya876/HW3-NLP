## Stakeholder Analysis

**15. Identify 4+ stakeholders whose interests stand to be impacted.**

1. **People who post content**, especially speakers of African American English (AAE) and other dialects. Their posts, reach, account status, and ability to express themselves can be affected.
2. **People targeted by hate speech or harassment.** They need a platform where threats and abusive content are addressed reliably.
3. **Content moderators and trust-and-safety teams.** They need useful tools that help prioritize harmful content without overwhelming them with false alarms.
4. **The platform/company.** It has interests in user safety, compliance with applicable rules, trust, reputation, and operating costs.
5. **Advertisers and other organizations whose content appears beside posts.** They may want their ads kept away from genuinely harmful content, while also wanting fair and predictable moderation.
6. **The wider public and communities represented on the platform.** They have an interest in fair participation, public discussion, and not having one community’s language unfairly treated as suspicious.

**16. For each stakeholder, identify 2+ interests or values at stake.**

- **People who post content:** freedom of expression and fair treatment; access to their audience; privacy and clear explanations when content is moderated. For example, someone writing a harmless joke in AAE should not lose visibility just because of dialect markers.
- **People targeted by hate speech:** safety and dignity; timely action against threats and harassment; confidence that reports will be taken seriously. A system that misses direct threats can leave people exposed to abuse.
- **Moderators:** accurate and understandable signals; manageable workloads; the ability to use context and exercise judgment. If many harmless posts are flagged, moderators spend time reviewing false alarms instead of urgent cases.
- **The platform:** user safety and trust; reliable enforcement; operational efficiency and compliance. A fast system is useful, but a pattern of unfair removals can damage trust.
- **Advertisers:** brand safety; predictable moderation; avoiding association with genuinely hateful content without relying on unfair treatment of particular communities.
- **Wider public and communities:** equal participation; protection from discrimination; a healthy public conversation in which different dialects are respected.

These interests can conflict. For example, removing posts quickly may protect targets of abuse, but automatically removing every post above a model-score threshold can silence people whose language the model misunderstands (Sap et al., 2019).

**17. Identify 3+ conflicts of interests or values. Explain how each arises.**

1. **Safety versus freedom of expression.** Removing content quickly can reduce exposure to threats and harassment, but an incorrect removal can prevent harmless speech. For example, a friendly AAE expression could be mistaken for an insult.
2. **Speed and scale versus context and due process.** Automated systems can screen huge volumes of posts, but they do not always understand conversation history, sarcasm, quoting, or whether a phrase is reclaimed by a community. Human review improves context but takes time and resources.
3. **Advertiser/platform risk management versus fair treatment of users.** A platform may want to minimize the chance that ads appear beside harmful content, but an overly cautious system may suppress legitimate posts from dialect-speaking users. A lower-risk business decision can still distribute the cost unfairly.
4. **Consistency versus cultural and linguistic diversity.** A single rule is easy to apply at scale, but language changes across communities and contexts. Treating one dialect’s surface features as a universal warning sign can produce unequal error rates.

The conflicts are not solved by choosing one value and ignoring the others. The system needs proportional responses, evidence of actual harm, and a way to correct mistakes.

# Task 4: Value Investigations

**18. For each of the 3 conflicts identified in question 17, explain how the trade-off should be approached.**

1. **Safety versus freedom of expression:** Prioritize preventing credible threats and targeted abuse, but do not treat a model score as proof that a post is harmful. For borderline cases, send the post for human review, explain the decision, and allow an appeal. A direct threat may justify urgent action; a harmless post that uses a dialect marker should not be removed on that basis alone.
2. **Speed and scale versus context and due process:** Use automation to sort and prioritize posts, not as the only decision-maker for severe penalties. For example, a high-confidence threat can be escalated immediately, while an ambiguous joke or quoted slur can be reviewed with its surrounding conversation. The platform should measure false positives and false negatives, including separately for relevant language groups.
3. **Business risk versus fair treatment:** Advertiser safety is a legitimate concern, but it does not justify discriminating against users because of their dialect. The platform should use content-based evidence, publish moderation rules, and audit whether its decisions impose unequal burdens. Revenue or convenience should not override basic fairness and a meaningful appeal process.

Across all three conflicts, the response should be proportional to the likely harm. The system should protect people from abuse while preserving harmless expression and treating users fairly (Sap et al., 2019).

**20. List any other harms you can think of (at least 2).**

1. **Chilling effects and self-censorship:** People may stop posting, avoid their usual dialect, or avoid discussing sensitive topics because they fear a false flag. For example, a creator might rewrite a harmless post into a less natural style to avoid being misclassified.
2. **Double bind:** A person may be criticized if they use their community’s normal language, but may also feel pressured to change their language to satisfy a system that favors standard phrasing. This places the burden of correcting the system on the user.
3. **Moderator fatigue and misplaced attention:** If many harmless posts are incorrectly flagged, moderators must review them and may have less time for credible threats. False positives can therefore harm both wrongly flagged users and people who need urgent protection.
4. **Loss of trust:** If users believe moderation is arbitrary or biased, they may stop reporting abuse or stop trusting platform decisions. That can make it harder for the platform to protect people effectively.

## Technical Investigations

**21. How might this technology be misused or abused, beyond the dialect bias measured?**

The system could be misused in several ways:

- **Censoring criticism or dissent:** A platform, government, or powerful organization could label criticism as “offensive” to reduce its visibility, even when it is not hate speech. For example, a post criticizing a company’s policy could be mass-reported and then automatically down-ranked.
- **Coordinated false reporting:** Groups could intentionally report an opponent’s posts to trigger automated review or account penalties. The model may appear to confirm the reports even when the context is harmless.
- **Overbroad surveillance or profiling:** If toxicity scores are combined with user histories, the platform might build profiles of people as “risky” based on their language or associations. That could affect them beyond the original moderation purpose.
- **Gaming the system:** People who want to harass others may deliberately misspell slurs or use coded language to avoid detection, while benign users are flagged for ordinary language.

These examples show that technical accuracy alone is not enough. Access controls, transparency, limits on how predictions may be used, and independent audits are also important.

**22. Propose a technical intervention (beyond dialect/race priming) to mitigate the bias observed or reduce a stakeholder conflict. Explain and justify it.**

I would implement a **dialect-aware evaluation and threshold-review pipeline** before deployment and during operation:

1. Build a carefully reviewed evaluation set with harmless and harmful examples across AAE and other dialects, including matched sentences that express the same meaning in different phrasing.
2. Measure false-positive rates, false-negative rates, precision, and recall for each group where it is appropriate and ethically possible to do so. Report uncertainty and sample sizes instead of relying only on one overall accuracy number.
3. If a group has a higher false-positive rate, inspect the examples and training labels, improve the data, and test the change on a separate held-out set. Do not simply lower the threshold for one group without understanding the causes and implications.
4. Use the model to prioritize human review for ambiguous cases, and reserve automatic severe penalties for narrowly defined, high-confidence cases with safeguards and appeal.

This intervention is justified by my Part 2 results. C1 showed that 95.37% of AAE-marker-flagged records were labelled offensive in the dataset, compared with 81.99% of records not flagged by the heuristic (ratio 1.163×). C2 showed a higher mini-model false-positive rate in the AAE-marker group (47.06% versus 25.61%), but that group had only 17 truly non-offensive test examples. D1’s ten matched pairs had mean offensive probabilities of 0.7419 versus 0.7036 (ratio 1.054×), while D2’s mean toxicity scores were 0.0072 versus 0.0018 (ratio 3.891×). The matched-pair probe has only ten hand-written pairs, and the models output different kinds of scores, so these values are warning signals for further testing rather than a definitive fairness verdict. This pipeline addresses the conflict between efficient moderation and fair treatment by measuring errors, correcting them, and escalating uncertain cases to people.

**23. Propose a regulation or platform policy to mitigate anticipated harms or reduce conflicts between stakeholders. Explain and justify it.**

I would require a **clear notice, meaningful appeal, and independent bias-audit policy**:

- Tell users when content has been removed, down-ranked, or penalized and provide a plain-language reason and a way to appeal.
- Do not suspend an account or impose a severe penalty based only on an uncertain toxicity score; provide human review for serious or disputed cases.
- Regularly audit false positives and false negatives across dialects and other relevant groups, publish aggregate results, and document corrective actions.
- Limit use of toxicity predictions to the moderation purpose for which they were collected; do not reuse them for unrelated profiling without a separate justification and safeguards.
- Keep records of moderation decisions so the platform can identify repeated patterns and correct systematic errors.

This policy balances safety with expression and due process. It also gives moderators feedback about where the model fails. As context, the European Commission explains that the EU Digital Services Act includes transparency obligations around certain content-moderation decisions and statements of reasons; the exact legal duties depend on the service and situation (European Commission, “How the Digital Services Act enhances transparency online”). The policy proposed here is a recommendation, not a claim that every listed measure is legally required in every country.

# Reflection

**24. What is the point of automated hate speech/offensive-content detection on a large platform?**

Its main purpose is to help a platform handle more posts than human moderators could inspect one by one, identify potentially harmful content quickly, and prioritize urgent cases. For example, a credible threat could be sent to a moderator sooner. The system should support human decision-making rather than replace it completely, because a prediction can be wrong or miss context. Sap et al. (2019) show why evaluation matters: models trained on existing labels can produce different false-positive rates across dialect groups.

**25. What values does this technology promote, and what values does it put at risk?**

It can promote **safety, well-being, timely response, and efficient use of moderator time** by helping identify harassment and threats. At the same time, it can put **fairness, freedom of expression, privacy, due process, and cultural/linguistic respect** at risk. For example, removing a credible threat can protect a target, but removing a harmless AAE post because of its wording can unfairly limit the author’s speech. The goal should be to protect people from real abuse without treating a dialect or identity as a substitute for evidence of harmful intent.

**26. Does dialect and race priming (as proposed in the paper) fully solve the problem measured in Part 2? Why or why not?**

No. Sap et al. (2019) found that dialect priming and race priming reduced offensive ratings in their annotation experiment; for “offensive to anyone,” the average rating fell from 0.55 in the control condition to 0.44 in both priming conditions (Figure 3). That is useful evidence that annotation instructions and social context can improve judgments, but it is not a complete solution.

My Part 2 results also show why continued testing is needed. The dataset’s offensive-label rate was 95.37% for AAE-marker-flagged records and 81.99% for records not flagged by the heuristic (C1 ratio 1.163×). The mini classifier’s false-positive rate was 47.06% versus 25.61% in C2, although the AAE-marker group included only 17 truly non-offensive test examples, making the estimate unstable. In the ten matched sentence pairs, D1’s average offensive probability was 0.7419 for AAE-style phrasing and 0.7036 for standard phrasing; D2’s average toxicity score was 0.0072 and 0.0018, respectively. These small probes suggest sensitivity to phrasing, but they cannot establish the size of bias in real-world use.

Priming mainly targets human annotation behavior. It does not automatically repair already-collected datasets, guarantee that every annotator understands every dialect, remove model shortcuts, or address changes in language over time. It should be one part of a broader approach that includes representative data, expert/community review, group-wise evaluation, human review, appeals, and ongoing monitoring.

**27. All things considered, should platforms deploy automated hate speech detection as described in the case study? If so, what regulation or technical fixes are needed?**

I would support **careful, limited deployment as a moderation aid**, but not unrestricted automatic removal or account suspension based only on a model score. Platforms need some scalable way to identify potential threats and harassment, yet my Part 2 results show that an overall accuracy of 94.17% does not tell the whole story. The mini classifier correctly identified many offensive examples, but its false-positive rate differed across the marker groups in C2; that estimate is uncertain because the AAE-marker group had only 17 truly non-offensive test examples. The matched-pair tests also showed higher average scores for AAE-style wording, but there were only ten pairs.

Before deployment, I would require testing on a larger, carefully reviewed set of matched examples and real examples across dialects; reporting false-positive and false-negative rates as well as overall accuracy; independent audits; human review for ambiguous cases and serious penalties; clear explanations and appeals; and periodic re-evaluation as language changes. Platforms should also examine the data labels themselves, because a model can reproduce patterns in human annotations. This approach aims to protect people targeted by hate speech while reducing unfair penalties against people whose language is misunderstood (Sap et al., 2019).
