# Task 1: Analyzing Representational and Allocative Harms

As a refresher:

**Representational harms** occur when a group is marginalized based on identity (e.g. race, gender, ethnicity, dialect), reinforcing negative associations, stereotypes, or erasing a group from visibility.

**Allocative harms** occur when resources, opportunities, or benefits (e.g. access to a platform, a job, a service) are distributed in a way that disproportionately disadvantages a group based on identity.

__Identify whether each example represents a representational harm, an allocative harm, or both. Provide reasoning for your classification.__

1. **Both representational and allocative harm.** Representational harm means the system gives people an unfair impression of a group. Here, it treats AAE (African American English) as if it is a sign that a post is offensive. Allocative harm means someone loses something valuable, such as visibility or access. If the platform hides the post because of the incorrect label, the writer loses an opportunity to reach readers. For evidence, Sap et al. (2019, Table 1) found a correlation of *r* = 0.42 between inferred AAE and the “offensive” label in the DWMW 17 dataset. Figure 2 reports that the model falsely flagged 46.3% of AAE examples in the offensive category, compared with 9.0% for White-aligned English. This shows that the errors were not evenly distributed.

2. **Primarily allocative harm, and it can also cause representational harm.** Down-ranking means showing a post to fewer people; removal means taking it away from the platform. Both reduce the writer’s access to an audience, so they are allocative harms. If harmless AAE posts are removed more often, the platform may also give the impression that AAE is inappropriate, which is representational harm. For example, a harmless joke written in AAE might be hidden while a similar joke in another dialect stays visible. Sap et al. (2019, Figure 2) found different false-positive rates across dialect groups. The paper did not directly test a platform’s down-ranking policy, but its results show how biased predictions could lead to unequal treatment if a platform acts on them.

3. **Both representational and allocative harm.** If annotators do not understand AAE and the instructions do not ask them to consider dialect or context, they may label a harmless expression as offensive. This misrepresents the speaker’s language, causing representational harm. If a model learns from those labels and later hides the speaker’s posts, it can also cause allocative harm. For example, an annotator might misunderstand a familiar in-group expression as an insult even when the post is not attacking anyone. Sap et al. (2019, Table 1) found that inferred AAE correlated with “offensive” labels in DWMW 17 (*r* = 0.420) and with “abusive” labels in FDCL 18 (*r* = 0.355). These correlations show a pattern in the datasets; they do not prove that every annotator or every label was biased.

4. **Both representational and allocative harm.** The repeated incorrect label suggests that the creator’s way of speaking is offensive, which can reinforce a stereotype. Losing monetization is allocative harm because the creator loses a chance to earn money. For example, a creator’s harmless comedy written in AAE could be incorrectly flagged, and the platform might then stop the creator from earning money from that content. The error-rate differences reported by Sap et al. (2019, Figure 2) make this a plausible risk. However, the researchers measured labels and model predictions, not creators’ income, so monetization loss is a possible consequence rather than a result directly measured in the paper.

5. How do these harms reinforce existing stereotypes and systemic barriers?

    - These harms can form a cycle. First, people creating the dataset may label some AAE expressions as offensive because they misunderstand the dialect or its context. The model learns from those labels and may then flag more AAE posts. If the platform acts on those flags, AAE speakers may lose visibility or have their posts removed more often. This can make the unfair belief that “AAE is more offensive” seem true, even though the system helped create that pattern. Sap et al. (2019, §3.1 and Figure 2) found associations between AAE and toxicity labels and showed that trained models repeated differences in error rates. The authors also explain that AAE is common among, but not used by all and only, African Americans. Therefore, dialect should not be treated as proof of a person’s race.

6. What are some of the potential impacts of these harms on the marginalized groups involved?

    - People may have harmless posts removed or shown to fewer users, lose followers or income, receive account penalties, or decide not to post because they fear being flagged. This is called self-censorship: people hold back what they want to say to avoid punishment. For example, a creator might stop using their usual dialect online because they worry the system will mistake it for offensive language. Biased flags can also waste moderators’ time by sending harmless posts for review instead of focusing on genuinely harmful content. Sap et al. (2019, Figure 2) found a 46.3% false-positive rate for AAE in the DWMW 17 offensive category, compared with 9.0% for White-aligned English. The paper measured model errors, not all these social or financial effects; those are possible consequences if platforms use the predictions to penalize users.

# Task 3: Empirical Investigations

12. Based on the paper, what training data is typically used to build hate speech detection models, and how were the "ground truth" toxicity/offensiveness labels for that data determined?

    - Sap et al. (2019) studied models trained on collections of Twitter posts that had already been labelled by human crowdworkers. Two examples are DWMW 17, from Davidson et al. (2017), and FDCL 18, from Founta et al. (2018). DWMW 17 used labels such as “hate speech,” “offensive,” and “none.” FDCL 18 used “hateful,” “abusive,” “spam,” and “none.” Multiple people judged each post: DWMW 17 had at least three annotations per tweet, and FDCL 18 used five workers per tweet. The researchers then used these labels as the answers the model should learn to predict. They are called “ground truth” for training, but they are still human judgments, not perfect or completely objective facts. For example, an annotator who does not understand an AAE expression might mark it offensive even if it is harmless in context. Evidence for this concern appears in Sap et al. (2019, Table 1): inferred AAE correlated with the “offensive” label in DWMW 17 (*r* = 0.420) and the “abusive” label in FDCL 18 (*r* = 0.355). These correlations show a pattern in the datasets, although they do not prove that every label was wrong.

13. In your own words, summarize the main empirical finding of Sap et al. (2019) regarding AAE and toxicity labels. What did they find when they looked at models trained on this data, and what intervention did they propose to reduce annotator bias?

    - The authors found that AAE-related words and expressions were more often associated with labels such as “offensive” or “abusive” in commonly used datasets. When models learned from those datasets, they also made more false-positive errors on AAE tweets. For example, Table 1 reports a correlation of *r* = 0.42 between AAE and the “offensive” label in DWMW 17, and *r* = 0.355 between AAE and “abusive” in FDCL 18. Figure 2 shows that the DWMW 17 model’s false-positive rate for AAE in the offensive category was 46.3%, compared with 9.0% for White-aligned English; for FDCL 18’s abusive category, the rates were 26.0% and 4.5%, respectively. The authors also applied their models to other Twitter datasets and found that toxicity predictions differed by dialect and by users’ self-reported race. To try to reduce bias at the data-collection stage, they tested **dialect priming** (telling annotators the likely dialect) and **race priming** (encouraging them to consider the author’s likely racial background). Figure 3 shows that the average rating for “offensive to anyone” fell from 0.55 in the control group to 0.44 in both priming groups. For “offensive to you,” race priming reduced the average from 0.33 to 0.25. In simple terms, when annotators were reminded to consider dialect and identity, they were less likely to call AAE tweets offensive. This suggests that annotation instructions and context can affect labels, and that datasets should be built with care (Sap et al., 2019, §§3–4).
